## 2. Configuration sécurisée (`.env`)

**Règle d'or : on ne met JAMAIS de mot de passe ou de clé API en dur dans le code.**

Étapes :
1. Copie `.env.example` en `.env` à la racine du repo.
2. Remplis tes propres identifiants (Supabase, PostgreSQL, MySQL, SQLite).
3. Le fichier `.env` est ignoré par git (voir `.gitignore`) — il ne sera jamais poussé sur GitHub.


```python
import shutil, os

# Exécute cette cellule une seule fois pour créer ton fichier .env local
if not os.path.exists("../.env"):
    shutil.copy("../.env.example", "../.env")
    print("Fichier .env créé à partir de .env.example. Édite-le avec tes vraies valeurs.")
else:
    print(".env existe déjà, rien à faire.")
```
